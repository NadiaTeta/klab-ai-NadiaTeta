## Load pre-trained word vectors
This downloads GloVe, a ready-made set of word vectors. Every word is stored as a list of 50 numbers, learned from Wikipedia and news articles. The first run downloads about 66 MB.

```python
# Cell 1: load pre-trained GloVe vectors (50 numbers per word, trained on Wikipedia + news)
import gensim.downloader as api
model = api.load("glove-wiki-gigaword-50")
```

## Words closest to "free"
This asks the model for the 5 words whose vectors are nearest to the vector for "free", i.e. the words used in the most similar contexts.

```python
# Cell 2: which words are closest in meaning to "free"?
model.most_similar("free", topn=5)
```

## Similarity scores
This compares two pairs of words using cosine similarity. A score near 1 means the words are very similar; a score near 0 means they are unrelated.

```python
# Cell 3: similarity scores (closer to 1 = more similar)
print("cheap vs affordable:", model.similarity("cheap", "affordable"))
print("cheap vs banana:", model.similarity("cheap", "banana"))
```

## The king - man + woman analogy
This starts from "king", takes away "man", adds "woman", and asks which word is closest to the result.

```python
# Cell 4: the analogy king - man + woman = ?
model.most_similar(positive=["king", "woman"], negative=["man"], topn=1)
```

## Does GloVe know SMS slang?
This checks whether some common SMS slang words from my spam dataset are in GloVe's vocabulary. True means the word has a vector; False means GloVe has never seen it.

```python
# Cell 5: does GloVe know SMS slang? True = known, False = unknown
for word in ["wif", "lar", "lor", "tkts"]:
    print(word, "->", word in model)
```
