```python
def normalise(values, minimum=None, maximum=None):
    """
    Scale a list of numbers into the 0–1 range.

    Args:
        values (list): The numbers to scale.
        minimum (float, optional): The lower bound to use for scaling.
        maximum (float, optional): The upper bound to use for scaling.

    Returns:
        list: The normalised values between 0 and 1.
    """
    if not values:
        return []

    if minimum is None:
        minimum = min(values)

    if maximum is None:
        maximum = max(values)

    if minimum == maximum:
        return [0.0 for _ in values]

    return [(value - minimum) / (maximum - minimum) for value in values]
```

```python
print(normalise([12, 18, 25, 40]))
print(normalise([5, 5, 5, 5]))
```

```python
def summarise_scores(scores):
    """
    Summarise a list of scores.

    Args:
        scores (list): A list of numeric scores.

    Returns:
        dict: A dictionary containing the count, mean, minimum,
        maximum, and number of scores greater than or equal to 0.8.
        For an empty list, count and above_threshold are 0, while
        mean, minimum, and maximum are None.
    """
    if not scores:
        return {
            "count": 0,
            "mean": None,
            "minimum": None,
            "maximum": None,
            "above_threshold": 0
        }

    return {
        "count": len(scores),
        "mean": sum(scores) / len(scores),
        "minimum": min(scores),
        "maximum": max(scores),
        "above_threshold": sum(score >= 0.8 for score in scores)
    }
```

```python
print(summarise_scores([0.65, 0.72, 0.81, 0.90]))
print(summarise_scores([0.95, 0.88, 0.76, 0.82, 0.91]))
print(summarise_scores([]))
```

```python
def safe_divide(numerator, denominator, default=0.0):
    """
    Safely divide two numbers.

    Args:
        numerator (float): The number to be divided.
        denominator (float): The number to divide by.
        default (float, optional): The value returned if division
            cannot be performed.

    Returns:
        float: The result of the division, or default if the
        denominator is zero or either input is not numeric.
    """
    try:
        return numerator / denominator
    except (ZeroDivisionError, TypeError):
        return default
```

```python
print(safe_divide(10, 2))
print(safe_divide(10, 0))
print(safe_divide("10", 2))
```

```python
data = np.array([
    [70, 80, 90, 85],
    [75, 85, 95, 88],
    [80, 90, 100, 92]
])

# A 3x4 array has 3 rows and 4 columns; transposing swaps the rows and columns.

print("Original shape:", data.shape)
print("Transpose shape:", data.T.shape)
```

```python
column_means = data.mean(axis=0)

# axis=0 calculates the mean separately for each column.

print("Column means:", column_means)
```

```python
centred_data = data - column_means

# NumPy broadcasting subtracts each column's mean from every value in that column.

print("Column-centred data:")
print(centred_data)
```

```python
import pandas as pd

df = pd.DataFrame({
    "student": ["Nadia", "Alice", "Brian", "Clara", "David", "Eva"],
    "score": [0.92, 0.78, 0.85, 0.67, 0.95, None],
    "epochs": [10, 12, 8, 15, 9, 11],
    "passed": [True, False, True, False, True, True]
})

# This DataFrame contains numerical, boolean, and text data, including a missing score.

print(df)
```

```python
# Inspect the structure, descriptive statistics, and missing values in the DataFrame.

print("DataFrame information:")
df.info()

print("\nDescriptive statistics:")
print(df.describe())

print("\nMissing values:")
print(df.isna().sum())
```

```python
high_scores = df[df["score"] >= 0.8].sort_values("score", ascending=False)

# Filter scores of at least 0.8, then sort the matching rows from highest to lowest.

print(high_scores)
```

```python
df["performance"] = df["score"] >= 0.8

# This derived column uses the score to create a Boolean pass/fail indicator.

print(df)
```
